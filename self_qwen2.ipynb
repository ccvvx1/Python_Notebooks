### 3.1 通过插件显示网页

```python
from IPython import display

display.IFrame(src="https://www.baidu.com", width=1600, height=800)
```

```python
from IPython.display import IFrame, HTML, display
import uuid

# 生成唯一 ID 标识 iframe
iframe_id = "my-iframe-" + str(uuid.uuid4())

# 嵌入百度页面
iframe = IFrame(src="https://www.baidu.com", width=1600, height=800)
display(iframe)

# 注入 JavaScript 尝试修改链接行为
display(HTML(f"""
<script>
  (function() {{
    const iframe = document.getElementById('{iframe_id}');

    // 监听 iframe 加载完成
    iframe.onload = function() {{
      try {{
        // 获取 iframe 内部文档（跨域时会失败）
        const iframeDoc = iframe.contentDocument || iframe.contentWindow.document;

        // 修改所有 target="_blank" 的链接
        const links = iframeDoc.querySelectorAll('a[target="_blank"]');
        links.forEach(link => {{
          link.target = "_self"; // 强制在当前 iframe 跳转
        }});
      }} catch (error) {{
        console.error('跨域限制，无法修改 iframe 内容:', error);
      }}
    }};
  }})();
</script>
"""))
```

```python
from flask import Flask, Response
import requests

app = Flask(__name__)

@app.route('/proxy/baidu')
def proxy_baidu():
    response = requests.get('https://www.baidu.com')
    modified_html = response.text.replace('target="_blank"', 'target="_self"')
    return Response(modified_html, mimetype='text/html')

if __name__ == '__main__':
    app.run(port=5000)
```

```python
import selenium.webdriver as webdriver
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options as FirefoxOptions
```

```python
def selenium_firefox_agent(driver_path="geckodriver"):
    firefox_driver_path = f'{driver_path}'
    user_agent = 'Mozilla'
    firefox_options = FirefoxOptions()
    firefox_options.add_argument('--headless')
    firefox_options.set_preference("general.useragent.override", user_agent)

    driver = webdriver.Firefox(
        # executable_path=firefox_driver_path,
        options=firefox_options
    )
    print('Scraper setup complete!')
    return driver
```

```python
crawler = selenium_firefox_agent()
crawler.get("https://www.baidu.com/")
source = crawler.page_source
print(crawler.title)
crawler.save_screenshot("google.png")  # 保存为图片
```
